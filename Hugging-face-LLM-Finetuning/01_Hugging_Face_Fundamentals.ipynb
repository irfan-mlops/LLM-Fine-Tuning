
# Hugging Face Authentication — Quick Revision

## 1. Login Methods

Hugging Face provides four common authentication methods.

| Method | Command / Function | Purpose |
|---|---|---|
| Web Login | `huggingface.co/login` | Login through browser |
| CLI Login | `hf auth login` | Authentication through terminal |
| Python Login | `login()` | Authentication in Python or Jupyter |
| Environment Token | `HF_TOKEN` | Authentication for scripts and deployments |

## 2. Python Authentication

The `huggingface_hub` library provides three login functions.

```python
from huggingface_hub import login, notebook_login, interpreter_login

login()              # Automatically selects the login interface
notebook_login()     # Uses the Jupyter Notebook login interface
interpreter_login()  # Uses the terminal login interface
```

## 3. CLI Authentication

```bash
# Authenticate with Hugging Face
hf auth login

# Verify the authenticated account
hf auth whoami

# Display stored authentication tokens
hf auth list

# Switch between stored tokens
hf auth switch

# Log out from Hugging Face
hf auth logout
```

## 4. Token Permissions

| Token Permission | Purpose |
|---|---|
| Read | Access and download authorized models and datasets |
| Write | Create repositories and upload or modify content |
| Fine-grained | Restrict access to selected resources and operations |

## 5. Key Concepts

- **Authentication:** Verifies the identity of a Hugging Face user.
- **Access Token:** A credential used for authentication.
- **Read Permission:** Allows access to authorized resources.
- **Write Permission:** Enables repository creation and uploads.
- **CLI Authentication:** Used through the command-line interface.
- **Python Authentication:** Used within Python scripts and notebooks.
- **HF_TOKEN:** Environment variable for token-based authentication.

## 6. Important Notes

- Public models can often be accessed without authentication.
- Private or gated repositories may require authentication and approval.
- Creating repositories requires appropriate write access.
- Never expose access tokens in notebooks or GitHub repositories.
- Prefer fine-grained tokens with the minimum required permissions.

---

**Quick Reminder:**

**Login Method = How you authenticate.**

**Token Permission = What you are allowed to do.**
